```python
import requests  
import json  
import sys  
  
def fetch_event_matches(event_key, api_key, output_file):  
    """  
    Fetch match results for an event using the TBA API and save to JSON file  
    """  
    # API endpoint for event matches  
    url = f"https://www.thebluealliance.com/api/v3/event/{event_key}/matches"  
      
    headers = {  
        'X-TBA-Auth-Key': api_key,  
        'Content-Type': 'application/json'  
    }  
      
    try:  
        # Make API request  
        response = requests.get(url, headers=headers)  
        response.raise_for_status()  
          
        # Parse JSON response  
        matches_data = response.json()  
          
        # Save to file  
        with open(output_file, 'w') as f:  
            json.dump(matches_data, f, indent=2)  
          
        print(f"Successfully saved {len(matches_data)} matches to {output_file}")  
          
    except requests.exceptions.RequestException as e:  
        print(f"Error fetching data: {e}")  
        sys.exit(1)  
    except json.JSONDecodeError as e:  
        print(f"Error parsing JSON: {e}")  
        sys.exit(1)  
  
# Usage example  
if __name__ == "__main__":  
    event_key = "2026orore"  # event key  
    api_key = "uqTThWSrIgK7D7M3ct9fnwfIrj9m7ZzuCjwsgWsHzMtRl2xRNIm8pEQXVhfwOsBv"  
    output_file = f"{event_key}_matches.json"  
      
    fetch_event_matches(event_key, api_key, output_file)
```
